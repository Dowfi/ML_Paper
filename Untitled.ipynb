```python
import numpy as np
import pandas as pd

np.random.seed(42)  # reproducibility

n = 1000

# Numeric features (sampled from normal distribution)
data = {
    "age": np.random.normal(0, 1, n),
    "watch_hours": np.random.normal(0, 1, n),
    "last_login_days": np.random.normal(0, 1, n),
    "monthly_fee": np.random.normal(0, 1, n),
    "number_of_profiles": np.random.normal(0, 1, n),
}

# Gender one-hot
genders = np.random.choice(["Male", "Other", "Female"], n, p=[0.45, 0.1, 0.45])
data["gender_Male"] = (genders == "Male").astype(int)
data["gender_Other"] = (genders == "Other").astype(int)
# implicit: Female = all zeros

# Subscription one-hot
subs = np.random.choice(["Premium", "Standard", "Basic"], n, p=[0.3, 0.5, 0.2])
data["subscription_type_Premium"] = (subs == "Premium").astype(int)
data["subscription_type_Standard"] = (subs == "Standard").astype(int)
# implicit: Basic = all zeros

# Payment method one-hot
payments = np.random.choice(["Crypto", "Debit Card", "Gift Card", "PayPal"], n)
data["payment_method_Crypto"] = (payments == "Crypto").astype(int)
data["payment_method_Debit Card"] = (payments == "Debit Card").astype(int)
data["payment_method_Gift Card"] = (payments == "Gift Card").astype(int)
data["payment_method_PayPal"] = (payments == "PayPal").astype(int)

# Convert to DataFrame
df_synthetic = pd.DataFrame(data)

print(df_synthetic.head())
print("Shape:", df_synthetic.shape)
```

```python
df_synthetic = df_synthetic.drop(columns=["churn_prob", "churn_risk"], errors="ignore")
```
